```python
import pandas as pd
import numpy as np

df = pd.read_csv(
    "/kaggle/input/datasets/nikitabisht/menstrual-cycle-data/FedCycleData071012 (2).csv"
)

# Replace blank strings
df = df.replace(r'^\s*$', np.nan, regex=True)

# Forward fill within each patient
df = (
    df.groupby("ClientID", group_keys=False)
      .apply(lambda x: x.ffill())
      .reset_index(drop=True)
)

print(df.shape)
df.head()
```

```python
import numpy as np

# Copy
cycle_df = df.copy()

# Convert numeric columns
numeric_cols = [
    "Age",
    "Height",
    "Weight",
    "BMI",
    "MeanCycleLength",
    "LengthofCycle"
]

for col in numeric_cols:
    cycle_df[col] = pd.to_numeric(cycle_df[col], errors="coerce")

# Sort by woman and cycle
cycle_df = cycle_df.sort_values(
    ["ClientID", "CycleNumber"]
)

# Previous cycle
cycle_df["Prev_Cycle_Length"] = (
    cycle_df.groupby("ClientID")["LengthofCycle"]
    .shift(1)
)

# Previous 2 cycles
cycle_df["Prev2_Cycle_Length"] = (
    cycle_df.groupby("ClientID")["LengthofCycle"]
    .shift(2)
)

# Rolling Mean (last 3 completed cycles)
cycle_df["Rolling_Mean_3"] = (
    cycle_df.groupby("ClientID")["LengthofCycle"]
    .transform(
        lambda x: x.shift(1).rolling(3).mean()
    )
)

# Rolling Std
cycle_df["Rolling_STD_3"] = (
    cycle_df.groupby("ClientID")["LengthofCycle"]
    .transform(
        lambda x: x.shift(1).rolling(3).std()
    )
)

print(cycle_df[[
    "ClientID",
    "CycleNumber",
    "LengthofCycle",
    "Prev_Cycle_Length",
    "Prev2_Cycle_Length",
    "Rolling_Mean_3",
    "Rolling_STD_3"
]].head(15))
```

```python
from sklearn.model_selection import GroupShuffleSplit

# Remove rows where lag features don't exist
model_df = cycle_df.dropna(
    subset=[
        "Prev_Cycle_Length",
        "Prev2_Cycle_Length",
        "Rolling_Mean_3",
        "Rolling_STD_3"
    ]
).copy()

# Features available BEFORE a cycle starts
FEATURES = [
    "Age",
    "Height",
    "Weight",
    "BMI",
    "CycleNumber",
    "Prev_Cycle_Length",
    "Prev2_Cycle_Length",
    "Rolling_Mean_3",
    "Rolling_STD_3"
]

TARGET = "LengthofCycle"

# Convert to numeric
for col in FEATURES + [TARGET]:
    model_df[col] = pd.to_numeric(
        model_df[col],
        errors="coerce"
    )

# Drop any remaining missing values
model_df = model_df.dropna(subset=FEATURES + [TARGET])

X = model_df[FEATURES]

y = model_df[TARGET]

groups = model_df["ClientID"]

print("="*60)
print("Dataset Shape :", model_df.shape)
print("Unique Women  :", groups.nunique())
print("Target Mean   :", y.mean())
print("Target Std    :", y.std())
print("="*60)

# Group split
gss = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_idx, test_idx = next(
    gss.split(
        X,
        y,
        groups
    )
)

X_train = X.iloc[train_idx]
X_test = X.iloc[test_idx]

y_train = y.iloc[train_idx]
y_test = y.iloc[test_idx]

print("Train:", X_train.shape)
print("Test :", X_test.shape)

print(
    "\nShared Women:",
    len(
        set(groups.iloc[train_idx]).intersection(
            set(groups.iloc[test_idx])
        )
    )
)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from catboost import CatBoostRegressor
from xgboost import XGBRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import pandas as pd
import numpy as np

models = {
    "Linear Regression": LinearRegression(),

    "Random Forest": RandomForestRegressor(
        n_estimators=300,
        random_state=42,
        n_jobs=-1
    ),

    "XGBoost": XGBRegressor(
        n_estimators=300,
        learning_rate=0.05,
        max_depth=5,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42
    ),

    "CatBoost": CatBoostRegressor(
        iterations=500,
        learning_rate=0.05,
        depth=6,
        loss_function="RMSE",
        random_seed=42,
        verbose=False
    )
}

results = []

predictions = {}

for name, model in models.items():

    print(f"Training {name}...")

    model.fit(X_train, y_train)

    pred = model.predict(X_test)

    predictions[name] = pred

    mae = mean_absolute_error(y_test, pred)

    rmse = np.sqrt(mean_squared_error(y_test, pred))

    r2 = r2_score(y_test, pred)

    results.append([
        name,
        mae,
        rmse,
        r2
    ])

results_df = pd.DataFrame(
    results,
    columns=[
        "Model",
        "MAE",
        "RMSE",
        "R2"
    ]
)

results_df = results_df.sort_values(
    "RMSE"
)

display(results_df)
```

```python
import os
import json
import pickle
import shutil

import joblib
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# =====================================================
# Create Folder Structure
# =====================================================

ROOT = "cycle_length_prediction_engine_v1"

MODELS = os.path.join(ROOT, "models")
REPORTS = os.path.join(ROOT, "reports")
IMAGES = os.path.join(ROOT, "images")

os.makedirs(MODELS, exist_ok=True)
os.makedirs(REPORTS, exist_ok=True)
os.makedirs(IMAGES, exist_ok=True)

# =====================================================
# Save Best Model
# =====================================================

best_model = models["Linear Regression"]

joblib.dump(
    best_model,
    os.path.join(MODELS, "cycle_length_model.pkl")
)

joblib.dump(
    FEATURES,
    os.path.join(MODELS, "feature_columns.pkl")
)

# =====================================================
# Metadata
# =====================================================

metadata = {
    "model_name": "Cycle Length Prediction Engine",
    "version": "v1",
    "algorithm": "Linear Regression",
    "target": TARGET,
    "features": FEATURES,
    "train_samples": int(len(X_train)),
    "test_samples": int(len(X_test)),
    "unique_users": int(groups.nunique())
}

with open(
    os.path.join(MODELS, "model_metadata.json"),
    "w"
) as f:
    json.dump(metadata, f, indent=4)

# =====================================================
# Predictions
# =====================================================

pred = best_model.predict(X_test)

pred_df = X_test.copy()
pred_df["Actual"] = y_test.values
pred_df["Predicted"] = pred
pred_df["Error"] = pred_df["Actual"] - pred

pred_df.to_csv(
    os.path.join(REPORTS, "predictions.csv"),
    index=False
)

# =====================================================
# Metrics
# =====================================================

mae = mean_absolute_error(y_test, pred)
rmse = mean_squared_error(y_test, pred) ** 0.5
r2 = r2_score(y_test, pred)

metrics_df = pd.DataFrame({
    "Metric": ["MAE", "RMSE", "R2"],
    "Value": [mae, rmse, r2]
})

metrics_df.to_csv(
    os.path.join(REPORTS, "model_metrics.csv"),
    index=False
)

# =====================================================
# Training Summary
# =====================================================

with open(
    os.path.join(REPORTS, "training_summary.txt"),
    "w"
) as f:

    f.write("Cycle Length Prediction Engine v1\n")
    f.write("="*50 + "\n\n")

    f.write(f"Algorithm : Linear Regression\n")
    f.write(f"Target    : {TARGET}\n")
    f.write(f"Train     : {len(X_train)}\n")
    f.write(f"Test      : {len(X_test)}\n")
    f.write(f"Users     : {groups.nunique()}\n\n")

    f.write(f"MAE  : {mae:.4f}\n")
    f.write(f"RMSE : {rmse:.4f}\n")
    f.write(f"R2   : {r2:.4f}\n")

# =====================================================
# Actual vs Predicted
# =====================================================

plt.figure(figsize=(8,6))
plt.scatter(y_test, pred)
plt.xlabel("Actual Cycle Length")
plt.ylabel("Predicted Cycle Length")
plt.title("Actual vs Predicted")
plt.grid(True)
plt.tight_layout()
plt.savefig(
    os.path.join(IMAGES, "actual_vs_predicted.png"),
    dpi=300
)
plt.close()

# =====================================================
# Residual Distribution
# =====================================================

residuals = y_test - pred

plt.figure(figsize=(8,6))
plt.hist(residuals, bins=30)
plt.xlabel("Residual")
plt.ylabel("Frequency")
plt.title("Residual Distribution")
plt.grid(True)
plt.tight_layout()
plt.savefig(
    os.path.join(IMAGES, "residual_distribution.png"),
    dpi=300
)
plt.close()

# =====================================================
# Residual Scatter
# =====================================================

plt.figure(figsize=(8,6))
plt.scatter(pred, residuals)
plt.axhline(0, linestyle="--")
plt.xlabel("Predicted")
plt.ylabel("Residual")
plt.title("Residual Scatter")
plt.grid(True)
plt.tight_layout()
plt.savefig(
    os.path.join(IMAGES, "residual_scatter.png"),
    dpi=300
)
plt.close()

# =====================================================
# Zip Everything
# =====================================================

shutil.make_archive(
    ROOT,
    "zip",
    ROOT
)

print("="*60)
print("Everything Saved Successfully!")
print("Folder :", ROOT)
print("ZIP    :", ROOT + ".zip")
print("="*60)
```
